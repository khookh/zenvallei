# Explore annual Zennevallei NDVI
Load the selected Sentinel-2 observations, inspect their quality and explore maps, distributions and pixel time series.

```python
import matplotlib.pyplot as plt
from greenwave_ndvi import discover_observations, open_observation, open_stack

inventory = discover_observations()
inventory[['year', 'date', 'cloud_affected_percentage', 'coverage_percentage']]
```

```python
scene = open_observation(2023)
scene.ndvi.plot(cmap='RdYlGn', vmin=-0.2, vmax=0.9, size=7)
plt.title(f"NDVI {scene.attrs['date']}")
plt.show()
```

```python
annual = open_stack(chunks=(1, 512, 512))
annual
```

```python
annual.ndvi.mean(('y', 'x'), skipna=True).compute().plot(marker='o')
plt.title('Mean valid NDVI by selected observation')
plt.grid(alpha=0.25)
plt.show()
```

```python
# Replace these projected coordinates with a location of interest.
sample = annual.ndvi.sel(x=annual.x.values[len(annual.x) // 2], y=annual.y.values[len(annual.y) // 2], method='nearest')
sample.compute().plot(marker='o')
plt.title('NDVI time series for one 10 m pixel')
plt.show()
```
